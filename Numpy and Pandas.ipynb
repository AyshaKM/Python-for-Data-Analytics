```python
import numpy as np
import pandas as pd
```

```python
#1. Create a 1D NumPy Array:

temperatures_w1 =np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("week 1 Temperatures:",temperatures_w1)
```

```python
#2. Inspection and Properties:

print("Shape:", temperatures_w1.shape)
print("Data type:", temperatures_w1.dtype)
print("Number of elements:", temperatures_w1.size)

print
```

```python
#3. Array Operations:

#Convert Celsius to Fahrenheit

temperatures_w1_f = (temperatures_w1 * 9/5) + 32

print("week 1 Temperatures in Fahrenheit:",temperatures_w1_f)



#Find the maximum, minimum, and mean
max_temp = np.max(temperatures_w1)
min_temp = np.min(temperatures_w1)
mean_temp = np.mean(temperatures_w1)

print("Maximum temperature:", max_temp)
print("Minimum temperature:", min_temp)
print("Mean temperature:", mean_temp)
```

```python
#4. Array Slicing and Indexing:

#first three days
first_three_days = temperatures_w1[:3]
print("First three days:", first_three_days)

#weekend
weekend_temps = temperatures_w1[[5, 6]]
print("Weekend temperatures:", weekend_temps)

#middle three days
middle_three_days = temperatures_w1[2:5]
print("Middle three days:", middle_three_days)
```

```python
#5. Create a 2D Array:

temperatures = np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
print(temperatures)
```

```python
#6. Inspect and Slice the 2D Array:

#Shape,data type and Number of elements

print("Shape:", temperatures.shape)
print("Data type:", temperatures.dtype)
print("Number of elements:", temperatures.size)
```

```python
#Extract temperatures of each week
week1 = temperatures[0]
week2 = temperatures[1]

print("Week 1 temperatures:", week1)
print("Week 2 temperatures:", week2)


#Extract temperatures for weekends
weekends = temperatures[:, [5, 6]]
print("Weekends temperatures:", weekends)
```

```python
#Pandas Series:
#1. Creating Pandas Series:

marks = pd.Series([95, 92, 89, 85, 80
],index = ['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'
])

print(marks)
```

```python
#2. Indexing and Slicing:

#access first rankusing integer index
print(marks.iloc[0])


#Retrieve top 3 ranks using loc
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

#retrieve 3rd rank using iloc
print(marks.iloc[2])

#Boolean mask - marks greater than 90
print(marks[marks > 90])
```

```python
#3. Manipulating Series:
 #modify rank1

marks.loc['Rank1'] = 100
print(marks)


#remove the last rank
marks = marks.drop('Rank5')
print(marks)


#calculate CGPA
CGPA = marks / 10
print(CGPA)
```

```python
#Pandas DataFrame:

#1. Creating Pandas DataFrame
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print(transactions)

```

```python
#2. Data Exploration:
print(transactions)

print(transactions.head())

print(transactions.tail())

print(transactions.shape)

print(transactions.columns)

print(transactions.dtypes)
```

```python
#display product category and amount
print(transactions[['ProductCategory', 'Amount']])
```

```python
#retrieve last 3 columns
print(transactions.iloc[:, -3:])
```

```python
#Filter north region and amount greater than 200

north_transactions = transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)]
print(north_transactions)
```

```python
#Value cunts for product category
product_category_counts = transactions['ProductCategory'].value_counts()
print(product_category_counts)
```

```python
#Find the unique values in the 'Region' column.
unique_regions = transactions['Region'].unique()
print(unique_regions)
```

```python
#Group by 'Region' and find the mean amount

mean_amount = transactions.groupby('Region')['Amount'].mean()
print(mean_amount)
```

```python
#3. Manipulating the DataFrame:

#modify 'Amount'

transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165
print(transactions)

```

```python
# adding discount column

transactions['Discount'] =   transactions['Amount'] * 0.10
print(transactions)
```

```python
#remove TransactiOnID 109
transactions = transactions[transactions['TransactionID'] != 109]
print(transactions)
```

```python
# Delete the 'Discount' column from the DataFrame.
transactions = transactions.drop(columns = ['Discount'])
print(transactions)
```
